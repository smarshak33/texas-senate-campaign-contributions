```python
trd_tx_100_names_clean = [name.strip().lower() for name in trd_tx_100_names]
```

```python
print(trd_tx_100_names_clean)
```

```python
# Create an empty list to store the rows we find
matched_dfs = []

# Loop through each titan and pull their data
for titan in trd_tx_100_names_clean:
  # Check if the name column contains the titan's name (case-insensitive)
  match = df[
      df["Contributor Name"].str.lower().str.contains(titan, na=False)
  ]
  matched_dfs.append(match)

# Combine all the matched entries into one isolated DataFrame
isolated_df = pd.concat(matched_dfs).drop_duplicates()

# Print how many rows you successfully found
print(f"Isolated {len(isolated_df)} contribution entries matching the TRD 100.")
```

```python
# Save to a clean CSV file
isolated_df.to_csv("/Users/samanthamarshak/dataProjects/texas-senate-campaign-contributions/trd_texas_contributions.csv", index=False)
```
